```python
import random

import numpy as np
import pandas as pd
```

# Read data

```python
TWO_DATA_FILE = True
```

```python
if TWO_DATA_FILE:
    train_data_path = ""
    test_data_path = ""
else:
    combined_data_path = ""
```

```python
if TWO_DATA_FILE:
    df_train = pd.read_csv(train_data_path, index_col=0)
    df_test = pd.read_csv(test_data_path, index_col=0)
else:
    combined_data = pd.read_csv(combined_data_path, index_col=0)
```

# Set Time Resolution
Choose one of: `Hourly`, `Daily`, `Weekly`, `Monthly`, `Quarterly`, or `Yearly`.

```python
time_resolution = "Hourly"
```

# Feature Extractor
The lags and rolling windows below follow Supplementary Table S1.

```python
def feature_extractor(data):
    data.loc[:, "y-1"] = data.loc[:, "y"].shift(1)
    data.loc[:, "y-2"] = data.loc[:, "y"].shift(2)
    data.loc[:, "y-3"] = data.loc[:, "y"].shift(3)

    if time_resolution == "Hourly":
        data.loc[:, "y-24"] = data.loc[:, "y"].shift(24)
        data.loc[:, "y-36"] = data.loc[:, "y"].shift(36)
        data.loc[:, "y-48"] = data.loc[:, "y"].shift(48)

        data.loc[:, "y-1_rolling_mean_4"] = data.loc[:, "y-1"].rolling(4).mean()
        data.loc[:, "y-1_rolling_std_4"] = data.loc[:, "y-1"].rolling(4).std()
        data.loc[:, "y-1_rolling_mean_12"] = data.loc[:, "y-1"].rolling(12).mean()
        data.loc[:, "y-1_rolling_std_12"] = data.loc[:, "y-1"].rolling(12).std()
        data.loc[:, "y-1_rolling_mean_24"] = data.loc[:, "y-1"].rolling(24).mean()
        data.loc[:, "y-1_rolling_std_24"] = data.loc[:, "y-1"].rolling(24).std()

        data.loc[:, "y-24_rolling_mean_4"] = data.loc[:, "y-24"].rolling(4).mean()
        data.loc[:, "y-24_rolling_std_4"] = data.loc[:, "y-24"].rolling(4).std()
        data.loc[:, "y-24_rolling_mean_12"] = data.loc[:, "y-24"].rolling(12).mean()
        data.loc[:, "y-24_rolling_std_12"] = data.loc[:, "y-24"].rolling(12).std()
        data.loc[:, "y-24_rolling_mean_24"] = data.loc[:, "y-24"].rolling(24).mean()
        data.loc[:, "y-24_rolling_std_24"] = data.loc[:, "y-24"].rolling(24).std()

    if time_resolution == "Daily":
        data.loc[:, "y-7"] = data.loc[:, "y"].shift(7)
        data.loc[:, "y-14"] = data.loc[:, "y"].shift(14)
        data.loc[:, "y-28"] = data.loc[:, "y"].shift(28)

        data.loc[:, "y-1_rolling_mean_4"] = data.loc[:, "y-1"].rolling(4).mean()
        data.loc[:, "y-1_rolling_std_4"] = data.loc[:, "y-1"].rolling(4).std()
        data.loc[:, "y-1_rolling_mean_7"] = data.loc[:, "y-1"].rolling(7).mean()
        data.loc[:, "y-1_rolling_std_7"] = data.loc[:, "y-1"].rolling(7).std()
        data.loc[:, "y-1_rolling_mean_28"] = data.loc[:, "y-1"].rolling(28).mean()
        data.loc[:, "y-1_rolling_std_28"] = data.loc[:, "y-1"].rolling(28).std()

        data.loc[:, "y-2_rolling_mean_4"] = data.loc[:, "y-2"].rolling(4).mean()
        data.loc[:, "y-2_rolling_std_4"] = data.loc[:, "y-2"].rolling(4).std()
        data.loc[:, "y-2_rolling_mean_7"] = data.loc[:, "y-2"].rolling(7).mean()
        data.loc[:, "y-2_rolling_std_7"] = data.loc[:, "y-2"].rolling(7).std()
        data.loc[:, "y-2_rolling_mean_28"] = data.loc[:, "y-2"].rolling(28).mean()
        data.loc[:, "y-2_rolling_std_28"] = data.loc[:, "y-2"].rolling(28).std()

    if time_resolution == "Weekly":
        data.loc[:, "y-4"] = data.loc[:, "y"].shift(4)
        data.loc[:, "y-52"] = data.loc[:, "y"].shift(52)

        data.loc[:, "y-1_rolling_mean_4"] = data.loc[:, "y-1"].rolling(4).mean()
        data.loc[:, "y-1_rolling_std_4"] = data.loc[:, "y-1"].rolling(4).std()
        data.loc[:, "y-1_rolling_mean_12"] = data.loc[:, "y-1"].rolling(12).mean()
        data.loc[:, "y-1_rolling_std_12"] = data.loc[:, "y-1"].rolling(12).std()

    if time_resolution == "Monthly":
        data.loc[:, "y-6"] = data.loc[:, "y"].shift(6)
        data.loc[:, "y-12"] = data.loc[:, "y"].shift(12)
        data.loc[:, "y-24"] = data.loc[:, "y"].shift(24)

        data.loc[:, "y-1_rolling_mean_3"] = data.loc[:, "y-1"].rolling(3).mean()
        data.loc[:, "y-1_rolling_std_3"] = data.loc[:, "y-1"].rolling(3).std()
        data.loc[:, "y-1_rolling_mean_6"] = data.loc[:, "y-1"].rolling(6).mean()
        data.loc[:, "y-1_rolling_std_6"] = data.loc[:, "y-1"].rolling(6).std()
        data.loc[:, "y-1_rolling_mean_12"] = data.loc[:, "y-1"].rolling(12).mean()
        data.loc[:, "y-1_rolling_std_12"] = data.loc[:, "y-1"].rolling(12).std()

        data.loc[:, "y-12_rolling_mean_3"] = data.loc[:, "y-12"].rolling(3).mean()
        data.loc[:, "y-12_rolling_std_3"] = data.loc[:, "y-12"].rolling(3).std()
        data.loc[:, "y-12_rolling_mean_6"] = data.loc[:, "y-12"].rolling(6).mean()
        data.loc[:, "y-12_rolling_std_6"] = data.loc[:, "y-12"].rolling(6).std()
        data.loc[:, "y-12_rolling_mean_12"] = data.loc[:, "y-12"].rolling(12).mean()
        data.loc[:, "y-12_rolling_std_12"] = data.loc[:, "y-12"].rolling(12).std()

    if time_resolution == "Quarterly":
        data.loc[:, "y-4"] = data.loc[:, "y"].shift(4)
        data.loc[:, "y-8"] = data.loc[:, "y"].shift(8)

        data.loc[:, "y-1_rolling_mean_2"] = data.loc[:, "y-1"].rolling(2).mean()
        data.loc[:, "y-1_rolling_std_2"] = data.loc[:, "y-1"].rolling(2).std()
        data.loc[:, "y-1_rolling_mean_4"] = data.loc[:, "y-1"].rolling(4).mean()
        data.loc[:, "y-1_rolling_std_4"] = data.loc[:, "y-1"].rolling(4).std()
        data.loc[:, "y-1_rolling_mean_8"] = data.loc[:, "y-1"].rolling(8).mean()
        data.loc[:, "y-1_rolling_std_8"] = data.loc[:, "y-1"].rolling(8).std()

        data.loc[:, "y-4_rolling_mean_2"] = data.loc[:, "y-4"].rolling(2).mean()
        data.loc[:, "y-4_rolling_std_2"] = data.loc[:, "y-4"].rolling(2).std()
        data.loc[:, "y-4_rolling_mean_4"] = data.loc[:, "y-4"].rolling(4).mean()
        data.loc[:, "y-4_rolling_std_4"] = data.loc[:, "y-4"].rolling(4).std()

    if time_resolution == "Yearly":
        data.loc[:, "y-4"] = data.loc[:, "y"].shift(4)
        data.loc[:, "y-5"] = data.loc[:, "y"].shift(5)

        data.loc[:, "y-1_rolling_mean_2"] = data.loc[:, "y-1"].rolling(2).mean()
        data.loc[:, "y-1_rolling_std_2"] = data.loc[:, "y-1"].rolling(2).std()
        data.loc[:, "y-1_rolling_mean_3"] = data.loc[:, "y-1"].rolling(3).mean()
        data.loc[:, "y-1_rolling_std_3"] = data.loc[:, "y-1"].rolling(3).std()
        data.loc[:, "y-1_rolling_mean_5"] = data.loc[:, "y-1"].rolling(5).mean()
        data.loc[:, "y-1_rolling_std_5"] = data.loc[:, "y-1"].rolling(5).std()

        data.loc[:, "y-2_rolling_mean_2"] = data.loc[:, "y-2"].rolling(2).mean()
        data.loc[:, "y-2_rolling_std_2"] = data.loc[:, "y-2"].rolling(2).std()
        data.loc[:, "y-2_rolling_mean_3"] = data.loc[:, "y-2"].rolling(3).mean()
        data.loc[:, "y-2_rolling_std_3"] = data.loc[:, "y-2"].rolling(3).std()

    data = data.dropna()
    data.index = np.arange(len(data))
    return data
```

Combine training and test rows before causal feature extraction. This lets each one-step test row use target observations from earlier time steps without using future targets.

```python
if TWO_DATA_FILE:
    combined_data = pd.concat([df_train, df_test], axis=0)
```

# Extract features

```python
new_data = feature_extractor(combined_data)
new_data.head()
```

# Save data

```python
new_data.to_csv("new_data.csv")
```
